## there other ways to unpack iterables, each with different results and methods

```python
l = [1,2,3,4,5,6,7]
```

```python
a,b = l[0],l[1:]
a,b
```

```python
a,*b = l
a,b
```

### they basiclly do the same thing, assign first list element to first variable, and the rest of the list to the second

### the advantage of using the second method, the method that has the sign * on It, is that It applies to any type, not just sequance types, aka list that numerate and coordinate each element

```python
a,*b = "hello gus"
a,b
```

### and of course It works this way with strings, as we said It works with all iterables
### but what so special about this method, is that It gives us a way to unpack non sequence types, aka dictionnarys and sets

```python
d = {"keyy":3, "key3":5,"key2":1}
s = {"sat","sata","brother","brothera"}
```

```python
a,b,c = d[0], d[1], d[2]
a,b,c,e = s[0], s[1], s[2], s[3]
# of course this wont work, sets and dictionnaries and sets dont have sequence, they get re-arranged on assignement anyways
# you cant use the iterable[int] method anyways lol
```

### and yeah we can use the classic method

```python
a,b,c = d
a,b,c
```

```python
a,b,c,e = s
a,b,c,e
```

### but this method is ultimetly better when required for specific use:

```python
a,*c = d
a,c
```

```python
*a,c = s
a,c
```
