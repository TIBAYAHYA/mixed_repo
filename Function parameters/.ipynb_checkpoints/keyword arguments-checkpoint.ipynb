## the star method, when called inside a function, is gonna force the inputer to use keyword argument however comes after the starred argument, otherwise results in an error

```python
def func1(a,b,*args):
    print(a,b,args)

func1(1,2,3,4)
```

```python
def func1(a,b,*args,d):
    print(a,b,args,d)

func1(1,2,3,4,5)
```

## error because d wasnt specified as a keywoard argument when calling the function

```python
def func1(a,b,*args,d):
    print(a,b,args,d)

func1(1,2,3,4,d = 5)
```

#### also worth noting that the starred argument is mandetory, meaning we can just not have nothing for It when we call the function

```python
def func1(a,b,*args,d):
    print(a,b,args,d)

func1(1,2,d = 4)
```

### we get an empty tuple because the argument arg found nothing left from the statement

### we can also modify our function to make It not accept positional arguments, and only work with keywoard arguments

```python
def func1(*,a):
    print(a)

func1(2)
```

### anything that comes after the star as an argument is supposedly going to be a keywoard argument statement, other wise we get an error as shown above, so code expected to be like this

```python
def func1(*,a):
    print(a)

func1(a = 2)
```

### as said, only what comes after the star argument, whats before It can be a positional argument

```python
def func1(a,b,*,c,d):
    print(a,b,c,d)

func1(1,2,c=3,d=4)
```
