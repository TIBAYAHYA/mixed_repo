## in addition to the star method *args, we dot have the real star method , the double star **kwargs method

### what this method essentially  does is scopes keyword arguments and store them in the form of a dictionary:
### the scooping works as below

```python
def func(**kwargs):
    print(kwargs)
func(1,b = 2,c = 3,d = 4,e = 5,f = 6)
```

### EVERYTHING given to the double star method must be a keyword argument
### failing to fulfill this condition will result in an errordef func(**kwargs):

```python
def func(**kwargs):
    print(kwargs)
func(a = 1,b = 2,c = 3,d = 4,e = 5,f = 6)
```

### and yes we can give It nothing, the rule is that It doesnt expect an argument, but whats given is supposed to be a keyword argument

### we can combine the args and kwargs arguments in a single use

```python
def func(*args,**kwargs):
    print(args)
    print(kwargs)
func(1,2,3,4,5,a= 9,b=8,e=7) 
```

### args will scoop the positional arguments, and kwargs will scoop keyword arguments, 
### and dont even think about It, positional arguments cannot come after keyword arguments so there no mixing in the arguments to make the program sort It
### It will just result in a simple error

```python
def func(*args,**kwargs):
    print(args)
    print(kwargs)
func(c = 22, 1,2,3,4,5,a= 9,b=8,e=7,0) 
```

### *args   results in a tuple ()
### *kwargs results in a dictionary {:}

### now for creating a function that illustrates most of what learnt

### function that calculates the high, low and average for a series of numbers

```python
def calc_hi_lo_average(*arguments, print_It = False):
    if not arguments:
        hi,lo,avg = 0,0,0
    else: 
        hi = max(arguments)
        lo = min(arguments)
        avg = sum(arguments)/len(arguments)
    if print_It:
        print("high = {0}, low = {1}, average = {2}".format(hi,lo,avg))
```

```python
calc_hi_lo_average(1,34,2462,2246,print_It = True)
```
