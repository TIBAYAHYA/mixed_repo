## what are packed values?

```python
(1,2,3,4,5) # tuples 
[1,2,3,4,5] # list
{1,2,3,4,5} # set
{1:2,3:4,5:6} # a dictionnary
"12345"     # string, yes even a string is a packed value
```

### so we can say that every iterable is considered a packed value

### what defines a tuple is not the parenthesis, rather It is the comma

```python
(1) # is just the integer 1
```

```python
(1,) # is the tuple that contains the integer 1
```

```python
a =  1, 2 ,3 ,4
b = ()
c = (1)
```

```python
print(type(a))
print(type(b))
print(type(c))
```

### an empty parenthesis can represent a tuple, but thats  the only exeption where a non comma makes a tuple


### so what is unpacking? unpacking is the action of splitting packed values into individual variables
### something like what there is below:

```python
a, b , c = [1,2,3] # very basic but effective nonetheless
# what happens is:
# a = 1
# b = 2
# c = 3
```

### one cool way to swap variables using unpacking is


```python
a,b = b,a
```

### why does this work? why doesnt b and a get the same value since we are assigning the value of a to b and then assigning the value of b to a?
### this works because python evaluates the entire right hand side before assiging to the left hand side

```python
s = {"h","i","a","v","b","b"}
for numba in s:
    print(numba)

# of course duplicate  on a set get removed, since sets can only contain unique values
```

```python
k1,k2,k3,k4,k5 = s
print(k1)
print(k2)
print(k3)
print(k4)
print(k5)
# same thing here, the set gets reordered and remove duplicates
```

### same thing with dictionaries as we seen with sets

```python
d = {"a":1,"k":2,"e":3}
```

```python
for kalo in d:
    print(kalo)
```

### we can iterate over keys, values or the item of a dictionnary

```python
for ele in d.values():
    print(ele)

for key,value in d.items():
    print(key,value)

for item in d.items():
    print(item)
```

### the method .items returns a tuple with two variables on origin, but we can set each of the two elements to two variables and there after get both the key and value as we please
