### same way we can use the star method for iterables unpacking, we can also use It for arguments, can come in handy If we are not sure of the lengh of the stuff or just want to write cleaner code

```python
listaa = 1,2,3,4,5,6,7
a,b,*c = listaa
a,b,c
print(listaa)
```

```python
def funct(a,b,*c):
    print(a,b,c)

funct(listaa)
```

### this results in an error, because the entire iterable is taken up as a single arguments
### what can we do smart head?
### use the unpacking method, unpack before using It as a parameter

```python
def funct(a,b,*c):
    print(a,b,c)

funct(*listaa)
#     ^this made the function work
```

### prefered to use the argument as *args , alot of people use It that way, and Its easier to understand

```python
def funct(a,b,*args):
    print(a,b,args)
listaa = ["bring","it","down","right","now","big","dog"]
funct(listaa)
```

### of course we get an error, cannot take a list as an argument for the entire thing, you need to unpack, Its so obvious yet you manage to screw It up

## a function that will try to find the average of numbers that are inside a list!

```python
def avg(*args):
    total = sum(args)
    lengh = len(args)
    if lengh == 0:
        return 0
    else : 
        return  total/lengh



numero_lista = [1,2,3,4,5,6,7,8,9]
print(avg(*numero_lista))
# cannot directly insert a list, have to unpack It
```

### How can this be useful? why cant we just use normal argumentation? well what If there is an infinite lengh of the list, we dont know how many arguments there are gonna be, then this comes in handy

### but this could get even smoother using the boolean tricks we previously learnt, we want to remove the if else statement as Its memory consuming and looks as ugly as your fat mother

```python
def avg(*args):
    total = sum(args)
    lengh = len(args)
    return lengh and total/lengh


numero_lista = [1,2,3,4,5,6,7,8,9]
print(avg(*numero_lista))
```

### otherwise we can force our function to have at least one positional argument!

```python
def avg(a, *args):
    total = sum(args) + a # plus the a
    lengh = len(args) +1 # plus the a as well
    return lengh and total/lengh
avg(2,2,3,5)
# may be better since If there is no argument the user gets an error
```
