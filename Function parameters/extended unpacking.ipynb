## there other ways to unpack iterables, each with different results and methods

```python
l = [1,2,3,4,5,6,7]
```

```python
a,b = l[0],l[1:]
a,b
```

```python
a,*b = l
a,b
```

### they basiclly do the same thing, assign first list element to first variable, and the rest of the list to the second

### the advantage of using the second method, the method that has the sign * on It, is that It applies to any type, not just sequance types, aka list that numerate and coordinate each element

```python
a,*b = "hello gus"
a,b
```

### and of course It works this way with strings, as we said It works with all iterables
### but what so special about this method, is that It gives us a way to unpack non sequence types, aka dictionnarys and sets

```python
d = {"keyy":3, "key3":5,"key2":1}
s = {"sat","sata","brother","brothera"}
```

```python
a,b,c = d[0], d[1], d[2]
a,b,c,e = s[0], s[1], s[2], s[3]
# of course this wont work, sets and dictionnaries and sets dont have sequence, they get re-arranged on assignement anyways
# you cant use the iterable[int] method anyways lol
```

### normal upacking, when applied to a dictionary, will only be applied to the keys of that dictionary, and to fix that, we have the double start method,
# **

```python
d1 = {"keyy":1, "key3":2, "key2":3}
d2 = {"keya":4, "keyt":5, "keyr":6}
d3 = {"goon":7, "koon":8, "keer":9}
the_big_d = **d1,**d2,**d3
the_big_d
# doesnt work because how are we supposed to represent a dictionary in the
```

```python
d1 = {"keyy":1, "key3":2, "key2":3}
d2 = {"keya":4, "keyt":5, "keyr":6}
d3 = {"goon":7, "koon":8, "keer":9}
the_big_d = {**d1,**d2,**d3}
the_big_d
```

### dictionarys can only contain unique keys, same with sets, If a new key is added that already exists on the dict, It overrates the already existing one with Its own key/value pair, or looks like It does nothing when It is a set

### with this operator, cool stuff can be done, like merging a dictionary while we are on the middle of writing It, and of course non unique get overriten

```python
d1 = {"keyy":1, "key3":2, "key2":3}
dd = {"keyy":3, "kratos":99 , "kamel":-20, **d1}
dd  # "keyy":1 overrote "keyy":3 since the later already existed on the dictionary
```
