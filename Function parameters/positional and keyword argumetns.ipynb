### in functions we can set a default parameter

```python
def that_function(a, b = 10):
    () #code
    
```

```python
def that_function(a, b = 10,c):
    () # code
```

### we cant define a function this way, the parameter/s we want to set as default must be the last on the function's tuple

```python
def that_function(a, b = 10, c = 30):
    print(a,b,c)
that_function(10)
that_function(5,2)
```

### what If we want a parameter on the middle of the tuple to take on Its default value?

```python
that_function(a = 5, c = 33)
#b takes on Its default value
```

### keyword argumentation also allows us to do some stuff, but they also put some restrictions on us, once we specify a keyword argument, every argumen after It on the tuple must also be keyword argument, otherwise program crashes

```python
that_function(c= 2, a = 55, b =  22)
```

```python
that_function(a=22, b = 10, 5)
```

```python
that_function(22, 19, 3) # positional arguments
that_function(a = 22, b = 19, c = 3) #keyword arguments
```

### both work, both have advantages and limitations
