# how to avoid errors when comparing stuff?

```python
a = 10
b = 0
if a/b :
    print("okay")
    # this results in an error, because we are trying to do a non possible operation,
    # there is a better way to handle this, see as below the error.
```

```python
# we can do It this way
a = 10
b = 0
if b :
    if a/b:  # if b is 0, then Its truthiness returns false
        print("okay")
else:
    print("cannot divide by zero my friend")
```

```python
# but short circuting has a finer way of doing It
a = 10
b = 0
if b and a/b:  # if b is 0, then Its truthiness returns false
    print("okay")
else:
    print("cannot divide by zero my friend")
```

```python
names = ["Abdul", "Ossama", "6ix9ine", "20cents", "Monkey","",""]
for name in names:
    if name[0] in string.digits:
        print(f"{name}, name cannot start with a digit")
```

### to avoid getting an error when the name[0] contains nothing, we can apply what we learnt

```python
names = ["Abdul", "Ossama", "6ix9ine", "20cents", "Monkey","",""]
for name in names:
    if name and name[0] in string.digits:
        print(f"{name}, name cannot start with a digit")
```
