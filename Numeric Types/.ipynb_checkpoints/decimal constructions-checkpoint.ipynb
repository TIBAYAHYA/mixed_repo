### precision attribute does not affect the decimal Itself, It only affect the operation, you can call It as an operation rounder

## we also know that global and local context are different
### wither we change the precision or another attribute, that only affects the inner decimals

```python
import decimal
from decimal import Decimal
a = Decimal(0.123456789)
b = Decimal(0.123456789)
print(f" operation: {a+b}")
with decimal.localcontext() as l_ctx:
    l_ctx.prec = 2
    c = a + b
    print(f"value of c in local context {c}")
print(f"value of c in global context {c}")
print(f"operation {a+b}")
```

### and of course value of c is same as It is on local context, since we assigned the result into a variable, the variable doesnt just get rid of It
