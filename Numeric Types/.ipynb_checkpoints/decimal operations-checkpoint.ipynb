# Float Devision and Modulo methods

```python
import decimal 
from decimal import Decimal
```

```python
x = 10
y = 3
divmod(x,y)
```

```python
 x//y
```

```python
x %y
```

```python
x == y * (x//y) +(x%y)
```

### divmod method is just the combinaison of both float division (//) and modulo(%) methods

```python
x = Decimal(-10)
y = Decimal(3)
divmod(x,y)
```

```python
x == y * (x//y) +(x%y)
```

### wither we deal with positive of negative numbers, decimals or not, the equation x == y * (x//y) +(x%y) is always satisfied
