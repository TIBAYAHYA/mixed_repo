```python
a = 0.3
b = 0.1 + 0.1 + 0.1
```

```python
a == b
```

### why is It like this? well simple look at this:

```python
format(a,"0.25f")
```

```python
format(b,"0.25f")
```

### even tho they are exactly float values,(3/10 fraction results in the exact float value 0.3 as opposed to 1/3 fraction that results in the 0.333333... non exact float value) the python floats approximation still messes theses things up

###  The above does work, but there is a more general and better way to sort this floats approximation problem, using something called ranged, and If two value are within that range, they are deemed equal:
#### technical explanation : if  |a-b|< the_range , the two values are deemed equal

```python
import math
a =  10000000000000.1
b =  10000000000000.2
```

```python
math.isclose(a,b, rel_tol = 1e-5, abs_tol = 1e-5)
a =  0.000000002
b =  0.000000001
```

```python
math.isclose(a,b, rel_tol = 1e-5, abs_tol = 1e-5)
```

#### this allows us to set a both a relative and an absolute tolerance to our float comparison, which gives us even more precision in the realm of unprecise python float 

```python
math.isclose(a,b,rel_tol = 1e-5, abs_tol = 1e-10)
```

#### we already know what each tolerance is based off, we play with the value to achieve our desired goal of the comparison
#### the isclose method can be more useful than normal comparison in the sense that It often takes the floats for what we actually meant

### coercing floats into an integer , something like 10.4, results in a certain data loss, we lose the float 0.4 off the value, and we can decide how this loss is going to be, do we get 10? 11?

```python
from math import trunc,floor,ceil
trunc(20.4)
```

```python
trunc(-20.4)
```

```python
int(20.4)
```

### the int method is basiclly the same as trunc method

```python
floor(12.5)
```

```python
floor(-12.5)
```

### the floor method returns the largest INTEGER less or equal than the float

```python
ceil(12.5)
```

```python
ceil(-12.5)
```

### the ceil method returns the smalled INTEGER greater or equal than the float

## trunc and floor might seem like the same thing, but there is a difference in negative numbers

```python
trunc(-10.4), trunc(-15.3)
```

```python
floor(-10.4), floor(-15.3)
```
