#### float objects can be written as fractions

```python
Fraction(3.75)
```

```python
Fraction(3.74)
```

#### This can be a helpful way to better represent floats, since python interpreter, and generally any interpreter at all has a limited precision when It comes to floats

```python
Fraction(355,113)
```

```python
355/113
```

## The fraction method can also take strings as parameters

```python
Fraction("10/2")   
```

```python
Fraction("10,2") 
```

### Although,for some reason the camma seperator on strings doesnt get interpreted as expected

```python
import math
x = Fraction(math.pi)
```

### we can limit the value of denominator,
### the method below means the closest fraction of x where the denominator doesnt exceed 10

```python
x.limit_denominator(10)
```

```python
22/7
```

```python
x.limit_denominator(100)
```

```python
311/99
```

### we can go further and further, and the further we go, the closer we get to real pi value

```python
x.limit_denominator(10000000)
```
