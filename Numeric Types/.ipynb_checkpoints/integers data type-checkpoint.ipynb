# size matters (yes It does)

```python
import sys

sys.getsizeof(2**1000)
```

```python
sys.getsizeof(None)
```

```python
sys.getsizeof(0)
```

```python
def multiply_by_two(n):
    for i in range(10000000): # 10 million times
        2*n        
```

```python
import time
start = time.perf_counter()
multiply_by_two(10)
end = time.perf_counter()
print(end-start)
```

```python
import time
start = time.perf_counter()
multiply_by_two(2**1000)
end = time.perf_counter()
print(end-start)
```

#### the sign // is for floor division, also known as div, put it as simple as just snaping the after comma part from existence

#### the sign % is the modulo operator, It also give the remainder of a division operation

```python
n = 155 // 4 #38
d = 155 % 4 #3
```

#### the thing about theses operators, is that when they are applied to same integer, they always end up in this relation below:


```python
n == d * (n//d) +(n%d) #True:
# where n is the numerator, and d is the denominator
```
