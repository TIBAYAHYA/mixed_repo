## how can rounding help?
#### lets say we want to find the average of three numbers:

```python
#one way would be to do this:
( 1+ 2 + 3 ) /3
```

```python
# or we can use the rounding method:
(round(1) + round(2) + round(3)) /3
```

### alot of times, this can be inaccturate, but It can be helpful when we want to find the average of a bunch of numbers without having to sum them up

## If we want the rounding to always prioritize getting away from zero, we gonna have to write a formula like this

#### sign(x) * int(abs(x)+0.5)
#### however there is no sign(x) method, the closest we can get to It is the math.copysign()method, which takes two arguments:
#### the first argument is the number that is gonna mirror the sign of second argument,
#### on our case, we use 1 as first argument, this way we multiply It by the result of absolute values methods, and we get to keep the sign as It is

```python
 # here is how It would look like If It was a function
def round_up(x):
    from math import copysign, fabs #fabs is the absolute method
    return copysign(1,x) * int(fabs(x) +0.5)


round_up(1.4)
```

```python
round_up(1.5)
```

```python
round_up(-1.5)
```

### basiclly prioritizes away from zero, in comparison to the normal rounding method, which prioritizes even significant digit
