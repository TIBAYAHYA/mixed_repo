### X or Y: if X is truthy, return X , otherwise evaluate Y and return Y

### X and Y: if X is falsly, return X , otherwise evaluate Y and return Y

# the or operator

```python
False or "okay"
```

```python
None or 1/0
```

### one cool way of setting a default value

```python
string1 = None
string2 = " "
string3 = "abcdefg"
string1 = string1 or "N/A"
string2 = string2 or "N/A"
string3 = string3 or "N/A"
print(string1,"\n",string2,"\n",string3)
```

```python
that_one_list = []
that_one_list = that_one_list or [0]
print(that_one_list)
```

### theses booleans operators allow us to set a default value for our objects, although there are certainly other ways to do It, but this is good to know

# the and operator

```python
# yeah, we can do this 
a = 2
b = 0
if b:
    print(a/b)
else:
    print(b)
```

```python
#but this is better
a = 2
b = 0
print(b and a/b)
```
