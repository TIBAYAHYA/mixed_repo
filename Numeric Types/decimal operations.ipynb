# Float Devision and Modulo methods

```python
import decimal 
from decimal import Decimal
```

```python
x = 10
y = 3
divmod(x,y)
```

```python
 x//y
```

```python
x %y
```

```python
x == y * (x//y) +(x%y)
```

### divmod method is just the combinaison of both float division (//) and modulo(%) methods

```python
x = Decimal(-10)
y = Decimal(3)
divmod(x,y)
```

```python
x == y * (x//y) +(x%y)
```

### wither we deal with positive of negative numbers, decimals or not, the equation x == y * (x//y) +(x%y) is always satisfied, both the methods are not floated division or modulo, again they just satisfy the previously stataed equation

# what else can we do with decimal type objects?

```python
a = Decimal("10.5")
print(a.sqrt())  # square root of  a
print(a.exp())   # exponent of  a
print(a.ln())    # natural logarithm of a
```

```python
import math
x = 0.01
dec_x = Decimal("0.01")
print(format(math.sqrt(x),"0.25f"))
print(format(math.sqrt(dec_x),"0.25f"))
print(format(dec_x.sqrt(),"0.25f"))
```

### the math module converts Decimal objects into integer or float object before starting the operation, and both methods' precision and the way they handle stuff differentiates, we dont get the same exact results
