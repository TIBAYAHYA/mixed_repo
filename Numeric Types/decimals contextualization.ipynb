### we can create a local context using this

```python
decimal.localcontext()  # this takes the default context as a parameter, on our case It is the global context
```

```python
type(decimal.localcontext())
```

### to use local context, we gonna need to use the width method, which takes care of the opening and closing of our local context on Its own

```python
with decimal.localcontext() as l_ctx: 
    print(type(l_ctx))
    print(l_ctx)
# worth noting that every time we create a local context, It takes the arguments of the default context, in most cases It is the global one
# unless you changed It
```
