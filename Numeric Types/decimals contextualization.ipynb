### we can create a local context using this

```python
decimal.localcontext()  # this takes the default context as a parameter, on our case It is the global context
```

```python
type(decimal.localcontext())
```

### to use local context, we gonna need to use the width method, which takes care of the opening and closing of our local context on Its own

```python
with decimal.localcontext() as l_ctx: 
    print(type(l_ctx))
    print(l_ctx)
# worth noting that every time we create a local context, It takes the arguments of the default context, in most cases It is the global one
# unless you changed It
```

### this allows allows to see the arguments of our local context, unlike what previously the local context method had shown us

```python
import decimal
from decimal import Decimal
a =  Decimal("1.25") # directly assigning floats doesnt work with this costume precision rounding, we gonna need to do It this way
b =  Decimal("1.35")
```

```python
with decimal.localcontext() as l_ctx: 
    l_ctx.prec = 22 # this will precise the number of significant digits that will be taken on our float object?(dunno)

    l_ctx.rounding = decimal.ROUND_HALF_UP # basiclly the string ROUND_HALF_UP, just for better clairty
    print(f"{l_ctx.prec},\n{decimal.getcontext()}")
    print(round(a,1))
    print(round(b,1))

print(round(a,1))
print(round(b,1))
# default global is half even, prioritizing the even non significant digit in case of distance being equal
# in local
```

### floats outside the with method get affected by the global context, and ones inside the with get affed by local context, simple really

### I guess the decimal.Decimal provides a more precise representation of floats huh?

### still same problem, why? because the float itself is stored as a float, meaning we didnt eliminate the non precise part('1.3500000000000000888178420')
### the solution to this is to pass the number in the form of a string, this way the non precise part gets eliminated
