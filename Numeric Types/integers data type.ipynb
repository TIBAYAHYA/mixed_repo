```python
def from_base10(n , b): # n for the number, b for the base
    if b < 2:
        raise ValueError("Base cannot be less than 2")
    elif n < 0:
        raise ValueError("The number cannot be less than 1")
    elif n == 0:
        return [0]

    digits = []
    while  n > 0:
        n,m = divmod(n,b) #this is same as m,n = n%b , n//b
        digits.insert(0,m) # insert the modulo as first element
    return digits
```

```python
from_base10(10,2)
```

```python
from_base10(255,16)
```
