# creation performance comparison

```python
from decimal import Decimal
def run_float(n):
    for i in range(n):
        a = 3.1415

def run_decimal(n):
    for i in range(n):
        a = Decimal("3.1415")
n = 10000000
```

```python
start = time.perf_counter()
run_float(n)
end = time.perf_counter()
time_taken_float = end - start
print(f"took {time_taken_float} to run the float function {n} times!")
```

```python

start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
time_taken_decimal = end - start
print(f"took {time_taken_decimal} to run the decimal function {n} times!")
```

```python
time_diff = time_taken_decimal / time_taken_float
print(f"the float function was {time_diff} times faster than the decimal function at creating the object {n} times")
```

# Methods performance comparison

```python
from decimal import Decimal
def run_float(n):
    a = 3.1415
    for i in range(n):
        a + a
        

def run_decimal(n):
    a = Decimal("3.1415")
    for i in range(n):
        a + a

n = 50000000
```

```python
start = time.perf_counter()
run_float(n)
end = time.perf_counter()
time_taken_float = end - start
print(f"took {time_taken_float} to run the float function {n} times!")
```

```python
start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
time_taken_decimal = end - start
print(f"took {time_taken_decimal} to run the decimal function {n} times!")
```

```python
time_diff = time_taken_decimal / time_taken_float
print(f"the float function was {time_diff} times faster than the decimal function at adding the objects {n} times!")
```

# Square root performance difference

```python
from decimal import Decimal
import math
def run_float(n):
    a = 3.1415
    for i in range(n):
        math.sqrt(a)
        

def run_decimal(n):
    a = Decimal("3.1415")
    for i in range(n):
        a.sqrt()

n = 5000000
```

```python
start = time.perf_counter()
run_float(n)
end = time.perf_counter()
time_taken_float = end - start
print(f"took {time_taken_float} to run the float function {n} times!")
```

```python
start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
time_taken_decimal = end - start
print(f"took {time_taken_decimal} to run the decimal function {n} times!")
```

```python
time_diff = time_taken_decimal / time_taken_float
print(f"the float function was {time_diff} times faster than the decimal function at squaring the objects {n} times!")
```
