```python
import decimal, math, sys, time
from decimal import Decimal
```

```python
a = math.pi
b = Decimal(str(math.pi))
```

```python
print(format(a,".25f"))
print(format(b,".25f"))
```

```python
size_a = sys.getsizeof(a)
size_b = sys.getsizeof(b)

print(f"size of float pi  : {size_a}")
print(f"size of decimal pi: {size_b}")
print(f"decimal pi occupys _{size_b/size_a}_ more space than float pi")
```

#### this is weird, float pi contains more significant digits, yet occupys less space than decimal pi

```python
from decimal import Decimal
def run_float(n):
    for i in range(n):
        a = 3.1415

def run_decimal(n):
    for i in range(n):
        a = Decimal("3.1415")
n = 1000000
```

```python
n = 10000000
start = time.perf_counter()
run_float(n)
end = time.perf_counter()
time_taken_float = end - start
print(f"took {time_taken_float} to run the float function {n} times!")
```

```python

start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
time_taken_decimal = end - start
print(f"took {time_taken_decimal} to run the decimal function {n} times!")
```

```python
time_diff = time_taken_decimal / time_taken_float
print(f"the float function was {time_diff} times faster than the decimal function at running the object {n} times")
```
